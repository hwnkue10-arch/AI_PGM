```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# 학습 데이터 생성
x_train = np.array([1,2,3,4,5], dtype=np.float32)
y_train = np.array([3,6,9,12,15], dtype=np.float32)

# 1. Sequential 모델 생성
model = Sequential()

# 2. Dense(1) 추가 (출력값 1개인 선형층)
model.add(Dense(1, input_shape=(1,)))

# 3. Optimizer와 Loss 설정
model.compile(
    optimizer='Adam',
    loss='mse'
)

# 4. 모델 학습
model.fit(
    x_train,
    y_train,
    epochs=300
)

# 테스트 데이터
x_test = np.array([6,7,8], dtype=np.float32)

# 예측
prediction = model.predict(x_test)

print("예측 결과:")
print(prediction)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense
from tensorflow.keras.datasets import mnist

# 1. 데이터 불러오기
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 2. 데이터 전처리 (0~255 → 0~1 정규화)
x_train = x_train / 255.0
x_test = x_test / 255.0

# 3. 모델 생성
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),
    Dense(10, activation='softmax')
])

# 4. 모델 컴파일
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# 5. 모델 학습
model.fit(
    x_train,
    y_train,
    epochs=5
)

# 6. 모델 평가
loss, accuracy = model.evaluate(x_test, y_test)

print("Loss :", loss)
print("Accuracy :", accuracy)

# 7. 예측
prediction = model.predict(x_test)

print("첫 번째 이미지의 예측 확률:")
print(prediction[0])

print("예측한 숫자:", prediction[0].argmax())
print("실제 숫자:", y_test[0])
```
