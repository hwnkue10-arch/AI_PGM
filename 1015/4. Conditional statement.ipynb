```python
try:
    a = int(input("정수 1: "))
    b = int(input("정수 2: "))

    if a == b:
        print("두 수는 같습니다.")
    else:
        result = a if a > b else b
        print(f"결과: {result}")

except ValueError:
    print("정수가 아닙니다")
```

```python
try:
    a = int(input("첫 번째 수를 입력하시오: "))
    b = int(input("두 번째 수를 입력하시오: "))
    c = int(input("세 번째 수를 입력하시오: "))

    largest = a
    if b > largest:
        largest = b
    if c > largest:
        largest = c

    print(f"가장 큰 수는 {largest} 입니다.")

except ValueError:
    print("정수가 아닙니다")
```

```python
height_cm = float(input("키를 입력하시오(cm): "))
weight_kg = float(input("몸무게를 입력 하시오(kg): "))

bmi = weight_kg / (height_cm / 100) ** 2

if bmi >= 25:
    status = "과체중"
elif bmi >= 18.5:
    status = "정상"
else:
    status = "저체중"

print(f"BMI {bmi:.1f}: {status}")
```

```python
try:
    month = int(input("값을 입력하세요: "))
    
    if 3 <= month <= 5:
        print("봄입니다.")
    elif 6 <= month <= 8:
        print("여름입니다.")
    elif 9 <= month <= 11:
        print("가을입니다.")
    elif month == 12 or 1 <= month <= 2:
        print("겨울입니다.")
    else:
        print("잘못 입력하셨습니다.")
except ValueError:
    print("잘못 입력하셨습니다.")
```

```python
try:  
    month = int(input("값을 입력하세요: "))  
    match month:
        case 3 | 4 | 5:
            print("봄입니다.")
        case 6 | 7 | 8:
            print("여름입니다.")
        case 9 | 10 | 11:
            print("가을입니다.")
        case 12 | 1 | 2:
            print("겨울입니다.")
        case _:
            print("잘못 입력하셨습니다.")
except ValueError:
    print("잘못 입력하셨습니다.")
```
