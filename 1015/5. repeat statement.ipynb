```python
nums = [3, 8, 2, 9, 5]
big = nums[0]
count = 0

for n in nums:
    if n > big:
        big = n
    if n % 2 == 1:
        count += 1

print(big, count)
```

```python
try:
    n = int(input("값을 입력하세요: "))
    if 0 < n < 51:
        for i in range(1, n + 1):
            if i % 3 == 0 and i % 5 == 0:
                print("FizzBuzz", end=" ")
            elif i % 3 == 0:
                print("Fizz", end=" ")
            elif i % 5 == 0:
                print("Buzz", end=" ")
            else:
                print(i, end=" ")
    else:
        print("입력값은 범위를 벗어났습니다")
except ValueError:
    print("입력값은 범위를 벗어났습니다")
```

```python
try:
    n = int(input("정수를 입력하세요: "))
    for i in range(1, n + 1):
        if n % i == 0:
            print(i, end=" ")
except ValueError:
    print("정수가 아닙니다")
```
