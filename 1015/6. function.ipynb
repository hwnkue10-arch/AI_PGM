```python
def discount_price(price, rate=10):
    sale = price * rate / 100
    return int(price - sale)

print(discount_price(20000))
print(discount_price(20000, 20))
```

```python
def convert_time(sec):
    m = sec // 60
    s = sec % 60
    return m, s

min, second = convert_time(125)
print(f"{min}분 {second}초")
```

```python
def introduce(name, age=20):
    print(f"안녕하세요, 제 이름은 {name}이고 {age}살입니다.")

user_name = input("이름을 입력하세요: ")
age_input = input("나이를 입력하세요: ").strip()

try:
    if age_input:
        introduce(user_name, int(age_input))
    else:
        introduce(user_name)
except ValueError:
    print("나이는 정수로 입력해주세요.")
```

```python
def check_even_odd(num):
    if num % 2 == 0:
        return "짝수"
    else:
        return "홀수"

input_num = input("정수를 입력하세요: ").strip()

try:
    num = int(input_num)
    result = check_even_odd(num)
    print(f"{num}는 {result}입니다.")
except ValueError:
    print("정수를 입력해주세요.")
```

```python
def is_passed(score):
    if score >= 80:
        return "합격"
    return "불합격"

user_score = input("점수를 입력하세요: ").strip()

try:
    score = int(user_score)
    result = is_passed(score)
    print(f"판정: {result}")
except ValueError:
    print("점수는 정수로 입력해주세요.")
```

```python
def find_max(a, b, c):
    return a if a > b and a > c else (b if b > c else c)

max_value = find_max(12, 45, 28)
print(f"가장 큰 수: {max_value}")
```

```python
def unique_sorted(lst):
    return sorted(set(lst))

unique_list = unique_sorted( [5, 2, 8, 2, 1, 5, 9])
print(unique_list)
```

```python
def sum_positives(lst):
    return sum(x for x in lst if x > 0)

list_values = [5, -2, 10, -7, 3]
positive_sum = sum_positives(list_values)
print(f"양수의 합: {positive_sum}")
```

```python
def count_letters(s):
    return sum(1 for char in s if char.isalpha())

letter_count = count_letters("Hello Python World!")
print(f"공백 제외 글자 수: {letter_count}")
```

```python
def get_ticket_type(age):
    if age < 13:
        return "어린이"
    elif age < 19:
        return "청소년"
    else:
        return "성인"
user_age = input("나이를 입력하세요: ").strip()
try:
    user_age = int(user_age)
    print(f"{user_age}세: {get_ticket_type(user_age)} 요금 적용")
except ValueError:
    print("나이는 정수로 입력해주세요.")
```

```python
def calc_bmi(weight, height):
    bmi = weight / ((height/100) ** 2)
    return bmi

weight = 70
height = 175
bmi = calc_bmi(weight, height)
print(f"BMI: {bmi:.1f}")
```

```python
def get_multiples_of_3(n):
    return [i for i in range(1, n+1) if i % 3 == 0]

input_number = input("정수를 입력하세요: ").strip()
try:
    n = int(input_number)
    multiples = get_multiples_of_3(n)
    print(f"3의 배수 리스트: {multiples}")
except ValueError:
    print("올바른 숫자를 입력하세요.")
```

```python
def is_palindrome(s):
    cleaned = ''.join(char.lower() for char in s if char.isalnum())
    return cleaned == cleaned[::-1]

word1 = "level"
word2 = "Python"
print(f"{word1}: {is_palindrome(word1)}/{word2}: {is_palindrome(word2)} ")
```
